```python
import numpy as np
import time

# دستگاه ۳۸۳
A = np.array([
    [5, 2, -1],
    [1, 6, -3],
    [2, 1, 4]
], dtype=float)

b = np.array([6, 4, 7], dtype=float)

#شرط توقف
eps = 1e-3
max_iter = 500

#تابع ژاکوبی
def jacobi(A, b, eps):
    n = len(b) #تعداد معادله
    x = np.zeros(n) #مفدار اولیه(صفر)
    x_new = np.zeros(n) #بردار مقدارهای جدید

    start = time.time() #برای مقایسه سرعت

    for iteration in range(1, 200): #حداقل ۲۰۰ تکرار

        x_new = np.zeros_like(x) #هر بار بردار جدید را خالی میکنیم

        for i in range(A.shape[0]): #حلقه برای محاسبه ژاکوبی
            s = 0
            for j in range(A.shape[1]): #سطر به سطر ماتریس(ضرایب دستگاه معادلات)
                if j != i:
                    s += A[i, j] * x[j] #جمع عناصر غیرقطری
            x_new[i] = (b[i] - s) / A[i, i] #استفاده از مقدار قبلی

        if np.linalg.norm(x_new - x) < eps: #اگر اختلاف دو تکرار بسیار کم باشد همگرا میشود
            end = time.time()
            return x_new, iteration, end - start

        x = x_new.copy() #انتقال مقدار جدید

    end = time.time()
    return x, max_iter, end - start


def gauss_seidel(A, b, eps):
    n = len(b)
    x = np.zeros(n)

    start = time.time()

    for iteration in range(1, 200):
        x_old = x.copy() #مقدار قبلی

        for i in range(A.shape[0]):
            s1 = sum(A[i, j] * x[j] for j in range(i)) #استفاده ار مقادیر جدید همین لحظه
            s2 = sum(A[i, j] * x_old[j] for j in range(i+1, n)) #استفاده از مقادیر قدیمی
            x[i] = (b[i] - s1 - s2) / A[i, i]

        if np.linalg.norm(x - x_old) < eps:
            end = time.time()
            return x, iteration, end - start

    end = time.time()
    return x, max_iter, end - start


x_j, itr_j, time_j = jacobi(A, b, eps)
x_g, itr_g, time_g = gauss_seidel(A, b, eps)

print("\n--- روش ژاکوبی ---")
print("جواب:", x_j)
print("تعداد تکرار:", itr_j)
print("زمان:", time_j)

print("\n--- روش گوس–سایدل ---")
print("جواب:", x_g)
print("تعداد تکرار:", itr_g)
print("زمان:", time_g)

print("\n==============================")
print("    مقایسه سرعت همگرایی")
print("==============================")
print(f"ژاکوبی:      {itr_j} تکرار | زمان = {time_j:.6f} ثانیه")
print(f"گوس‌سایدل:  {itr_g} تکرار | زمان = {time_g:.6f} ثانیه")
```
