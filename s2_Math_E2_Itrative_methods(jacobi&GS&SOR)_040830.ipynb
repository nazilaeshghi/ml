```python
import numpy as np
import time

# --------------------------
#   Jacobi Method
# --------------------------
def jacobi(A, b, tol=1e-6, max_iter=500):
    n = len(b)
    x = np.zeros(n)
    D = np.diag(A)
    R = A - np.diagflat(D)

    iterations = 0
    start = time.time()

    for _ in range(max_iter):
        x_new = (b - np.dot(R, x)) / D
        iterations += 1

        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            break
        x = x_new

    end = time.time()
    return x_new, iterations, end - start

# --------------------------
#   Gauss–Seidel Method
# --------------------------
def gauss_seidel(A, b, tol=1e-6, max_iter=500):
    n = len(b)
    x = np.zeros(n)
    iterations = 0
    start = time.time()

    for _ in range(max_iter):
        x_new = np.copy(x)
        iterations += 1

        for i in range(n):
            sum1 = np.dot(A[i, :i], x_new[:i])
            sum2 = np.dot(A[i, i+1:], x[i+1:])
            x_new[i] = (b[i] - sum1 - sum2) / A[i, i]

        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            break

        x = x_new

    end = time.time()
    return x_new, iterations, end - start

# --------------------------
#   SOR Method
# --------------------------
def sor(A, b, w=1.3, tol=1e-6, max_iter=500):
    n = len(b)
    x = np.zeros(n)
    iterations = 0
    start = time.time()

    for _ in range(max_iter):
        x_new = np.copy(x)
        iterations += 1

        for i in range(n):
            sum1 = np.dot(A[i, :i], x_new[:i])
            sum2 = np.dot(A[i, i+1:], x[i+1:])
            x_new[i] = x[i] + w * ((b[i] - sum1 - sum2) / A[i, i] - x[i])

        if np.linalg.norm(x_new - x, ord=np.inf) < tol:
            break

        x = x_new

    end = time.time()
    return x_new, iterations, end - start


# --------------------------
# Device System
# --------------------------
A = np.array([
    [-3, 1, 0, 3],
    [1, 6, 1, 0],
    [0, 1, 6, 1],
    [3, 0, 1, -3]
], dtype=float)

b = np.array([1,1,1,1], dtype=float)

# جواب واقعی :
x_real = np.array([-6, 1, 1, -6], dtype=float)

# --------------------------
# Run methods
# --------------------------
x_j, it_j, t_j = jacobi(A, b)
x_g, it_g, t_g = gauss_seidel(A, b)
x_s, it_s, t_s = sor(A, b)

# --------------------------
# Print results + errors
# --------------------------
print("Jacobi:", x_j, "iters:", it_j, "time:", t_j, "error:", np.linalg.norm(x_j - x_real))
print("Gauss-Seidel:", x_g, "iters:", it_g, "time:", t_g, "error:", np.linalg.norm(x_g - x_real))
print("SOR:", x_s, "iters:", it_s, "time:", t_s, "error:", np.linalg.norm(x_s - x_real))
```
